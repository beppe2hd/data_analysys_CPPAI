```python
temp = 20

if temp < 18:
  print("Fa freddo")
elif temp > 28:
  print("fa caldo")
else:
  print("clima ideale")
```

```python
temp = 18
tem_lim_sup = 28
tem_lim_inf = 18

if temp <= tem_lim_inf:
  print("Fa freddo")
elif temp >= tem_lim_sup:
  print("fa caldo")
else:
  print("clima ideale")

print("temperatura verificata")
```

```python
temp = 18
tem_lim_sup = 28
tem_lim_inf = 18
piove = False

if temp <= tem_lim_inf and piove == True:
  print("Non esco")
elif temp >= tem_lim_sup:
  print("fa caldo")
else:
  print("clima ideale")

print("temperatura verificata")
```

```python
temp = 23
tem_lim_sup = 28
tem_lim_inf = 18
piove = True

if temp <= tem_lim_inf:
  print("Fa freddo")
  if piove == True:
    print("Non esco")
  else:
    print("Posso uscire anche se fa feddo")
elif temp >= tem_lim_sup:
  print("fa caldo")
else:
  print("clima ideale")

print("temperatura verificata")
```

Classe CORRENTISTA

```python
class Correntista():

  def __init__(self, new_name, new_surname, new_balance, new_invested):
    self.name = new_name
    self.surname = new_surname
    self.balance = new_balance
    self.invested = new_invested

  def deposit(self, deposit_value):
    self.balance = self.balance + deposit_value
    print(f"The new amount is {self.balance}")

  def withdrawal(self, withdrawal_value):
    if withdrawal_value > self.balance:
      print(f"La massima somma che puoi ritirare é {self.balance}")
    else:
      self.balance = self.balance - withdrawal_value
      print(f"New balance is {self.balance}")

corr1 = Correntista('Maria', 'Rossi', 12500, 5000)
corr2 = Correntista('Vito', 'Ruggeri', 8600, 1000)
corr3 = Correntista('Silvia', 'Giannotta', 34800, 10000)
corr4 = Correntista('Cicco', 'Pasticcio', 500, 0)
```

```python
list_of_corr = [corr1, corr2, corr3, corr4]

for current_corr in list_of_corr:
  if current_corr.invested>0:
    print(f"{current_corr.name} {current_corr.surname} has invested {current_corr.invested}")
```

```python
corr3.balance
```

```python
corr3.deposit(500)
```

```python
corr3.deposit(100)
```

```python
corr3.withdrawal(50000)
```

```python
corr3.withdrawal(200)
```
